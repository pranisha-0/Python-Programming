```python
#q1 : sum of list elements
lst = []
n = int(input("Enter a number: "))
for i in range(1, n+1):
    num = int(input(f"Enter num{i}: "))
    lst.append(num)
print(f"Your list is: {lst}.")
sum = 0
for i in range(0, n):
    sum = sum + lst[i]
print(f"The sum of the list is: {sum}.")

```

```python
#Q2 : max and min in list
ls = []
n = int(input("ENter a number: "))
for i in range(0, n):
    nm = int(input(f"Enter num{i+1}: "))
    ls.append(nm)
print(f"Your list is {ls}")

max = ls[0]
for i in range(0, n):
    if ls[i] > max:
        max = ls[i]
print(f"THe max num from list is {max}.")

min = ls[0]
for i in range(0, n):
    if ls[i] < min:
        min = ls[i]
print(f"The min num from list is {min}.")
```

```python
#q3 : count no of even and odd nums from list
list = []
c_e = 0
c_o = 0
n =int(input("ENter a number: "))
for i in range(0, n):
    numr = int(input(f"ENter num{i+1}: "))
    list.append(numr)
for i in range(0, n):
    if list[i] % 2 == 0:
        c_e += 1
    else:
        c_o += 1
print(f"no. of even numbers is {c_e} & no. of odd numbers is {c_o}.")
```

```python
#q4 : reverse a list

normal_ls = []
reverse_ls = []
n = int(input("Enter a num: "))
for i in range(0, n):
    numb = int(input(f"ENter num{i+1}: "))
    normal_ls.append(numb)
for i in range(n, 0, -1):
    reverse_ls.append(normal_ls[i-1])
print(f"Your normal list is {normal_ls}.")
print(f"Your reversed list is {reverse_ls}.")
```

```python
#q5 : factorial using function
def factorial(a):
    if a == 0 and a == 1:
        return 1
    else:
        f = 1
        for i in range(1, a+1):
            f = f*i
        return f

n = int(input("ENter a number: ")) 
if n>0:
    fact = factorial(n)
    print("The factorial of", n, "is", fact)
else:
    print("Factorial of negative number doesn't exist.")
```

```python
#q6 : prime checker
def is_Prime(x):
    count = 0
    for i in range(1, x+1):
        if x%i == 0:
            count += 1
    if count == 2:
        return True
    else:
        return False

n = int(input("Enter a number: "))
check = is_Prime(n)
print(check)

```

```python
#q7 : remove duplicates
my_list = []
n = int(input("Enter a number: "))
for i in range(0, n):
    mem = input(f"Enter element{i+1}: ")
    my_list.append(mem)
no_dupes = []
for i in my_list:
    if i not in no_dupes:
        no_dupes.append(i)
print("original list:", my_list)
print("no duplicates:", no_dupes)

```

```python
#q8 : count occurrence of elements in list
my_list = []
n = int(input("Enter a number: "))
for i in range(0, n):
    mem = input(f"Enter element{i+1}: ")
    my_list.append(mem)
done = []
for i in range(0, n):
    count = 0
    for j in my_list:
        if my_list[i] == j:
            count += 1
    if my_list[i] not in done:
        print(f"frequency of {my_list[i]} -> {count}")
        done.append(my_list[i])
```
