```python
#q1
lst = []
n = int(input("Enter a number: "))
for i in range(1, n+1):
    num = int(input(f"Enter num{i}: "))
    lst.append(num)
print(f"Your list is: {lst}.")
sum = 0
for i in range(0, n):
    sum = sum + lst[i]
print(f"The sum of the list is: {sum}.")

```

```python
#Q2
ls = []
n = int(input("ENter a number: "))
for i in range(0, n):
    nm = int(input(f"Enter num{i+1}: "))
    ls.append(nm)
print(f"Your list is {ls}")

max = ls[0]
for i in range(0, n):
    if ls[i] > max:
        max = ls[i]
print(f"THe max num from list is {max}.")

min = ls[0]
for i in range(0, n):
    if ls[i] < min:
        min = ls[i]
print(f"The min num from list is {min}.")
```
