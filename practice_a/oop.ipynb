```python
class BankAccount:
    def __init__(self, acc_num, acc_holder, balance):
        self.a_n = acc_num
        self.a_h = acc_holder
        self.bal = balance
    def deposit(self, amt):
        self.bal += amt
        return (f"{amt} DEPOSITED.")
    def withdraw(self, amt):
        if amt > self.bal:
            print("Insufficient balance")
        else: 
            self.bal -= amt
            return (f"{amt} WITHDRAWN.")
    def disp_balance(self):
        return (f"CURRENT BALANCE: {self.bal}")

p1 = BankAccount(12422, "Pran", 10000)
print(p1.deposit(3000))
print(p1.disp_balance())

```

```python
class Humans:
    def __init__(self, name, age, gender):
        self.n = name
        self.a = age
        self.g = gender
    def intro(self):
        return f"{self.n} is a {self.a} y/o {self.g}."

class Student(Humans):
    def __init__(self, name, age, gender, school):
        Humans.__init__(self, name, age, gender)
        self.s = school

class Teacher(Humans):
    def __init__(self, name, age, gender, sub):
        Humans.__init__(self, name, age, gender)
        self.sub = sub

h = Humans("Prani", 18, "Female")
print(h.intro())

s = Student("Sushu", 20, "Female", "purwanchal")
print(s.intro())

t = Teacher("RLC", 67, "Male", "Maths")
print(t.intro())
```

```python
class Complex:
    def __init__(self, real, imag):
        self.r = real
        self.i = imag

    def __add__(self, c):
        return Complex(self.r + c.r, self.i + c.i)

    def input(self, n):
        self.r = float(input(f"Enter real part of {n}: "))
        self.i = float(input(f"Enter imaginary part of {n}: "))

    def __str__(self):
        return f"({self.r}+{self.i}j)"


c1 = Complex(0, 0)
c2 = Complex(0, 0)

c1.input("c1")
c2.input("c2")

c3 = c1 + c2

print(f"{c1} + {c2} = {c3}")
```
