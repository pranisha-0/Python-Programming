```python
#tuple
a = (1, 2, 3, 4, 5)
print(a[0])
print(a[-2])

#tuple unpacking
student = ("Pranisha", 17, "BEL", 4.0)
#student = name, age, dept, gpa #pack vaues into one tuple
name, age, dept, gpa = student  #unpack one tuple into variables
print(f'Unpacked -> Name: {name}, Age: {age}, Dept: {dept}, GPA: {gpa}')
```

```python
my_list = []
n = int(input("Enter a number: "))
for i in range(0, n):
    num = int(input(f"Enter number {i+1}: "))
    my_list.append(num)
my_tuple = tuple(my_list)
print(my_tuple)
```

```python
#sets
#inspired my mathemactical union sets etc
#mutable, unique collection of items, no duplicate values, unordered- indexing and slicing not possible
companies = {'Lacoste', 'Ralph Lauren'}
tech_companies = ['apple', 'google', 'apple']

# using update() method
companies.update(tech_companies)

print(companies)

a = {1, 2, 2, 3, 4, 5}
b = {4, 5, 6, 7, 6}
print(a)
print(a.union(b))
print(a.intersection(b))
print(b.difference(a))
print(a.symmetric_difference(b))
print(b.issubset(a))


c = "programming"
print(c)
chars = set(c)
print(len(chars))
print(chars)
```
