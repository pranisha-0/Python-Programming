```python

# Program 4.5.2 - Tic-Tac-Toe Board Simulation 
# Create empty 3x3 board 
board = [[' ' for _ in range(3)] for _ in range(3)] 
def print_board(b): 
    print('  0   1   2') 
for i, row in enumerate(b): 
    print(i, ' | '.join(row)) 
    if i < 2: 
        print(' -----------') 
def make_move(b, row, col, player): 
    if b[row][col] == ' ': 
    b[row][col] = player 
    return True 
    return False 
def check_winner(b): 
# Check rows and columns 
    for i in range(3): 
        if b[i][0] == b[i][1] == b[i][2] != ' ': 
            return b[i][0] 
        if b[0][i] == b[1][i] == b[2][i] != ' ': 
            return b[0][i] 
    # Check diagonals 
    if b[0][0] == b[1][1] == b[2][2] != ' ': 
        return b[0][0] 
    if b[0][2] == b[1][1] == b[2][0] != ' ': 
        return b[0][2] 
    return None 
 #Simulate a game 
moves = [('X', 0, 0), ('O', 1, 1), ('X', 0, 1), 
('O', 2, 0), ('X', 0, 2)] 
for player, r, c in moves: 
    make_move(board, r, c, player) 
print_board(board) 
winner = check_winner(board) 
print('\nWinner:', winner if winner else 'No winner yet')
```
