```python
#q1: given a list of numbers, use map() to find the squares
my = [1, 2, 3, 4, 5]
sqr = list(map(lambda x: x**2, my))
print("the squares of the numbers are:", sqr)
```

```python
#q2: convert a list of temperatures in celsius to fahrenheit using map()
cel = [0, 37, 100]
print("the temperatures in celsius are:", cel)
far = list(map(lambda x: (x * 9/5) + 32, cel))
print("the temperatures in fahrenheit are:", far)
```

```python
#q3: filter numbrs greater than 50 from a list
l = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print("the numbers in the list are:", l)
great = list(filter(lambda x: x > 50, l))
```

```python
#q4: filter all strings whose length is greater than 5
stri = ['anu', 'omu', 'pranisha', 'sushmita', 'karu', 'aashish']
gr_5len = list(filter(lambda x: len(x)>5, stri))
print(gr_5len)
```

```python
#q6: merge two dictionaries into one
d1 = {
    'a': 1,
    'b': 2
}
d2 = {
    'c': 3,
    'd': 4,
    'e': 5
}
merged = {**d1, **d2}
print(merged)
```

```python
#q7: separate even and odd num from tw list and store in new
l1 = [1, 2, 3, 4, 5]
l2 = [6, 7, 8, 9, 10]
merg = l1 + l2
even = list(filter(lambda x: x%2 == 0, merg))
odd = list(filter(lambda x: x%2 != 0, merg))
print("odd list: ", odd)
print("even list: ", even)
```

```python
#q8
def large(x):
    largest = max(x)
    return largest
def small(x):
    smallest = min(x)
    return smallest
def sum(x):
    total = sum(x)
    return total
def avg(x):
    average = sum(x)/len(x)
    return average
 
lst = [2, 4, 6, 7, 4, 3, 88, 39, 33, 20]
l = large(lst)
s = small(lst)
t = sum(lst)
a = avg(lst)
print("largest number is:", l)
print("smallest number is:", s)
print("sum of the numbers is:", t)

print("average of the numbers is:", a)
```
